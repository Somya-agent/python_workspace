```python
def count_words(filePath):
    """This function """
    count_dict={}
    with open(filePath,'r') as file:
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('!.:,?"\-')
                count_dict[word]=count_dict.get(word,0)+1
    return count_dict
```

##Fibonacci Print

```python
x=4
def fib(x):
    if(x==1 or x==0):
        return 1
    else:
        return fib(x-1)+fib(x-2)
    
print(fib(x))
```

#Palindrome

```python
s1="malayalam"

def palin(s1, start, end):
    
    if(start>=end):
        return True
    
    if(s1[start] != s1[end]):
        return False

    return palin(s1, start+1, end-1)
    
    

print(palin(s1, 0, len(s1)-1))

    
    
    
```

## Subsequences

```python
s1='abc'

def subsequencing(s):
    if( s==''):
        return ['']
    
    sub=subsequencing(s[1:])
    mychar=s[0]
    ans=[]
    ans.extend(sub)

    for eachPer in sub:
        ans.append(mychar+eachPer)

    return ans
    



print(subsequencing(s1))
```
